# Dataset

```python
datas_dir = "../../datas/raw"
filename = "2025_07_10_16.18.40.mp4"
dir_output_img = f"../../datas/preprocessing/{filename}"
```

```python
from stratadl.core.preprocessing.video import split_to_frames
from stratadl.core.utils.shards import create_shards_from_directory, make_dataloader, verify_distribution
from stratadl.core.utils.common import get_device

import torchvision.transforms as T

import logging
logging.basicConfig()
logging.getLogger().setLevel(logging.DEBUG)
```

```python
device = get_device()
device
```

```python
# On découpe la video en une série d'images
split_to_frames(datas_dir, dir_output_img)
```

```python
create_shards_from_directory(dir_output_img, '../../datas/preprocessing/wds', shards_max_size=100)
```

```python
train_wards_pattern = "../../datas/preprocessing/wds/train-*.tar"
val_wards_pattern = "../../datas/preprocessing/wds/val-*.tar"
test_wards_pattern = "../../datas/preprocessing/wds/test-*.tar"

img_size = (100,100)
```

```python
# Déclaration du pipeline de préprocessing
preproccessing = T.Compose([
    T.Resize(img_size),
    T.ToTensor() # Normalisation entre 0 et 1
])
```

```python
verify_distribution(train_wards_pattern, transform=preproccessing)
```

```python
verify_distribution(val_wards_pattern, transform=preproccessing)
```

```python
verify_distribution(test_wards_pattern, transform=preproccessing)
```

```python
train_loader = make_dataloader(train_wards_pattern, transform=preproccessing)
val_loader   = make_dataloader(val_wards_pattern, transform=preproccessing)
test_loader  = make_dataloader(test_wards_pattern, transform=preproccessing)
```
